# OpenCog Medical Demo - Interactive Walkthrough

Welcome to the OpenCog medical diagnostic demonstration. In this notebook, we'll see how **AtomSpace** (a hypergraph representation) and **PLN** (Probabilistic Logic Networks) work together to perform uncertain medical reasoning.

```python
# Setup the environment (if running in Colab)
import sys
import os
sys.path.append('..') # Ensure src is in path

from src.pln_reasoning.queries import diagnose_patient, explain_diagnosis, list_all_diseases
from src.atomspace_setup.knowledge_base import MedicalKnowledgeBase
```

## 1. Inspecting the Knowledge Base
First, let's load our AtomSpace.

```python
kb = MedicalKnowledgeBase().build()
print(f"Total Atoms in AtomSpace: {kb.atomspace.size()}")
print("Sample diseases:", list_all_diseases()[:5])
```

## 2. Running a PLN Diagnostic Query
We will run backward chaining on a set of symptoms: **Fever, Cough, and Fatigue**.

```python
symptoms = ["Fever", "Cough", "Fatigue"]
results = diagnose_patient(symptoms, top_k=3)

for r in results:
    print(f"Disease: {r.disease:12} | Score: {r.score*100:.1f}% | TruthValue: {r.tv}")
```

## 3. Explaining the AI's Decision
Symbolic AI is highly interpretable. Let's ask PLN *why* it ranked the top disease.

```python
best_disease = results[0].disease
explanation = explain_diagnosis(best_disease, symptoms)

print(f"Matched Symptoms: {explanation['matched_symptoms']}")
print("\nRule Chain:")
for rule in explanation['rule_chain']:
    print(f"  - {rule}")
```
